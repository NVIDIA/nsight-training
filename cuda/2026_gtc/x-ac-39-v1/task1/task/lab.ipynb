## Python GPU Frameworks

This lab covers profiling kernels from three major Python GPU frameworks. All of them compile down to CUDA/PTX, which means **Nsight Compute can profile them just like native C++ CUDA code**.
Nsight Compute operates at the CUDA level — it profiles **any** kernel that runs on the GPU, regardless of whether it was written in C++, Python/Triton, or Python/Numba. The profiler sees the same GPU instructions and can collect the same metrics.
