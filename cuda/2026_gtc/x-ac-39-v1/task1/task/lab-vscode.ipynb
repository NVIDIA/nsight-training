## Debugging Features

The `Numba CUDA` debugging solution supports a large number of features, not all of which will be covered in this lab. These features include:

* Linux support
* `C` / `C++` / `Fortran` debugging support
* `Numba CUDA` debugging support (beginning with CUDA Toolkit version r13.2)
* Mixed source language debugging. For example: you can debug calls to CUDA `C++` code from `Numba CUDA` kernels.
* Variable inspection and modification
* Execution control (`Run`, `Step Over`, `Step Into`, `Step Out`, etc)
* Setting breakpoints in the Graphical User Interface
* Setting breakpoints programmatically in Numba CUDA code by inserting `breakpoint()` calls
* Support for `Numba CUDA` polymorphic variables

Nsight Visual Studio Code Edition relies on CUDA GDB to provide the debugging engine, so all of the above operations can also be done from the CUDA GDB command-line. However, this lab will focus on debugging through the Visual Studio Code Graphical User Interface (aka VSCode).

## Components

### Numba and Numba CUDA

[Numba](https://numba.pydata.org/) is a just-in-time (JIT) compiler that translates Python functions into optimized machine code using the industry-standard LLVM compiler library. It enables writing code in a subset of the Python language, freeing the developer to implement custom algorithms without being constrained to only using pre-compiled library calls.

[Numba CUDA](https://nvidia.github.io/numba-cuda/) is based on Numba and provides the capability to write CUDA kernels for execution on the GPU. To compile a function to execute on the GPU instead of the CPU, use the `@numba.cuda.jit` decorator, which we will cover in more detail in the next section.

### Visual Studio Code and Nsight Visual Studio Code Edition

<img src="debugging/images/vscode-intro.png">

`Visual Studio Code` (VSCode) is a free, open-source, cross-platform code editor developed by Microsoft.
It is widely used for coding in many languages, including Python, C++, etc. It also supports debugging programs through its graphical user interface.

`Nsight Visual Studio Code Edition` is an extension for Visual Studio Code which enables debugging of CUDA programs, including support for Numba CUDA. This can be found and installed by clicking on the Extensions icon on the left-hand side of the VSCode window and searching for it by name.

This extension has been pre-installed for the lab - there's nothing to do here to install or enable it.

### CUDA GDB

CUDA GDB is NVIDIA's CUDA command-line debugger tool on Linux. Based on the industry-standard GDB debugger with extensions that support heterogeneous debugging of both C / C++ code running on the host CPU and CUDA C / C++ / Numba CUDA code running on an NVIDIA GPU.

CUDA GDB is a powerful tool for debugging, allowing you to set breakpoints in the code, inspect variables, and step through the code.

While primarily a command-line tool, CUDA GDB is integrated with Nsight Visual Studio Code Edition to act as the "debug engine" for the VSCode graphical user interface. This allows for a much more intuitive debugging environment, with far less of a learning curve than the CUDA GDB command-line interface.
