# Numba CUDA Debugging Fundamentals

The CUDA GDB and Nsight Visual Studio Code Edition debugger tools consist of the following executables:
* **`code`** — The Visual Studio Code integrated development environment, including the Nsight debugging extension.
* **`cuda-gdb`** — The NVIDIA CUDA GDB debugger executable

CUDA GDB and Visual Studio Code are integrated with each other to provide a rich debugging experience. Visual Studio Code provides a graphical user interface for the developer to interact with the debugger, while CUDA GDB provides the low-level debugging functionality.

In this lab, we'll use these components to debug a Numba CUDA kernel.

## Enabling Debugging in Numba CUDA

In Numba CUDA, GPU code is written in Python functions that are decorated with `@numba.cuda.jit`. These functions are compiled to CUDA PTX code at runtime, and then executed on the GPU. This decorator takes a number of optional arguments, including the signature of the function, the debug flag, and the optimization flag.

### Full Symbolic Debugging

To enable full symbolic debugging, set `debug=True` and `opt=False` in the `@numba.cuda.jit` decorator for the kernel. In this mode, you can step through the code line by line, insert and remove breakpoints, and inspect or modify variables. This is the primary mode for debugging Numba CUDA kernels.

### Line Number Only Debugging

Line number only debugging can be enabled by setting `lineinfo=True` and leaving the debug flag either unset or set to `False`.
In this mode, you can step through the code, but not inspect or modify variables.
Unlike full debugging support provided by the `debug=True` flag, line number only debugging does not require setting `opt=False`, but at the cost of not having variable or type information available.

Note that with optimizations enabled, the generated code can be very different from when optimizations are disabled, possibly resulting in unexpected behavior when debugging optimized code. This can include:

* Stepping through the code may appear to "jump around". For example, when stepping over a line of code, the program may appear to jump back to a previous line of code for a few instructions before continuing to make forward progress.
* Latent bugs that were hidden when running non-optimized code may be exposed.
